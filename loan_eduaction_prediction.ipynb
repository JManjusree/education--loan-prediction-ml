```python
# Title : Education Loan Prediction
#import Libraries
import random as r
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,confusion_matrix
#Dataset 

course_type=[
    "Engineering",
    "MBA",
    "BPhramacy",
    "Law",
    "Medicine",
    "Degree",
    "MTech"
             ]
d=[]
for i in range(1000):
    courses=r.choice(course_type)
    applicant_income=r.randint(10000,1000000)
    coapplicant_income=r.randint(10000,1000000)
    loan_amount=r.randint(400000,1500000)
    cibil_score=r.randint(500,900)
    cgpa=round(r.uniform(5.5,9.5),1)
    previous_score=r.randint(50,95)
    if(cgpa>=7.0 
       and cibil_score>=700 
       and previous_score>=70 
       and (applicant_income+coapplicant_income)>=50000):
        loan_status="Y"
    else:
        loan_status="N"

    d.append([courses,
              applicant_income,
              coapplicant_income,
              cibil_score,
              cgpa,
              previous_score,
              loan_amount,
              loan_status])
column=["Course",
        "Applicant Income",
        "Coapplicant Income",
        "Cibil Score",
        "Cgpa",
        "Previous Score",
        "Loan Amount" 
        ,"Loan Status"]
data=pd.DataFrame(d,columns=column)
data.to_csv("Dataset.csv",index=False)
print("dataset created successfully")




    
```

```python
#  Load DataSet
df=pd.read_csv("Dataset.csv")
df.head
```

```python
df.isnull().sum()
#Train-Test split
X=df.drop(columns=["Loan Status","Course"])
y=df["Loan Status"]
s=StandardScaler()
x=s.fit_transform(X)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
#Model Training and Model Evaluation
lg=LogisticRegression()
lg.fit(x_train,y_train)
lg_pred=lg.predict(x_test)
lg_accuracy=accuracy_score(y_test,lg_pred)


dt= DecisionTreeClassifier(
    criterion='gini',
    max_depth=5,
    random_state=42
)

dt.fit(x_train, y_train)
dt_pred = dt.predict(x_test)
dt_accuracy=accuracy_score(y_test,dt_pred)

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(x_train, y_train)
rf_pred = rf.predict(x_test)
rf_accuracy=accuracy_score(y_test,rf_pred)


#Model comparsion
results={"Logisitic Regression: ":lg_accuracy,
         "Decision Tree ":dt_accuracy,
         "Random Forest":rf_accuracy}
print("----Model Comparsion---")
for model,acc in results.items():
    print(f"{model}:{acc}")

print("The Best Model for this dataset is Decision Tree")
```
